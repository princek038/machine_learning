```python
from sklearn.preprocessing import PowerTransformer
```

```python
x=np.array([[-1],[0],[3],[5],[10],[50]])
```

```python
x
```

```python
pt=PowerTransformer(method="yeo-johnson")
```

```python
x_transformed=pt.fit_transform(x)
print("Original:")
print(x.ravel())
print("\nTranformed:")
print(np.round(x_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Department': ['Engineering', 'Data Science', 'HR', 'Engineering', np.nan, 'Finance', 'Data Science', np.nan],
    'City': ['New York', 'London', np.nan, 'Paris', 'New York', 'London', 'Paris', 'London'],
    'Education': ["Bachelor's", "Master's", np.nan, "Ph.D.", "Bachelor's", np.nan, "Master's", "Bachelor's"],
    'Age': [28, 34, np.nan, 42, 25, 39, np.nan, 30],
    'Salary': [65000, 85000, 50000, np.nan, 55000, 92000, np.nan, 70000],
    'Experience_Years': [2.5, 6.0, 1.0, np.nan, 1.5, 8.0, np.nan, 4.2]
}
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
from sklearn.impute import  SimpleImputer
```

```python
imputer=SimpleImputer(strategy="mean")
imputer.fit_transform(df[["Salary"]])
```

```python
imputer=SimpleImputer(strategy="mean")
df["new_salary"]=imputer.fit_transform(df[["Salary"]]).flatten()
df
```

```python
imputer=SimpleImputer(strategy="median")
df["new_Experience_Years"]=imputer.fit_transform(df[["Experience_Years"]]).flatten()
df
```

```python
imputer=SimpleImputer(strategy="most_frequent" ,)
df["City"]=imputer.fit_transform(df[["City"]]).flatten()
df
```
