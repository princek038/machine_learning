```python
array=np.linspace(12,22,num=15,dtype=int)
```

```python
def variance_std(*args):
    total=0
    for i in args:
        total+=(i-np.mean(args)**2)
        calculated_variance=total/(len(args)) 
        return calculated_variance ,calculated_variance**(1/2)   
```

```python
variance_std(*array)
```

```python
def s_variance_std(*args):
    total=0
    for i in args:
        total+=(i-np.mean(args)**2)
        calculated_variance=total/(len(args)-1) 
        return s_variance_std,calculated_variance**(1/2)   
```

```python
s_variance_std(*array)
```

```python
hours=[1,2,3,4,5,6]
marks=[30,40,45,55,65,75]
```

```python
import matplotlib.pyplot as plt
```

```python
# create a scatter plot using matplotlib
plt.figure(figsize=(4,4))
plt.scatter(hours,marks,marker="+",color="g")
plt.xlabel("nummber of hours")
plt.ylabel("marks")
plt.title("hours vs marks")
plt.show()
```
